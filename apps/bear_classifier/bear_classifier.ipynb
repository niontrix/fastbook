# A simple Bear Classifier

Upload a picture or photo of a bear and let it be classified as a grizzly, a black bear or a teddy bear by our amazing Bear Classifier Model.

```python
from fastai.vision.all import *
from fastai.vision.widgets import *
```

```python
path = Path()
learner = load_learner(path/"export.pkl")
upload_btn = widgets.FileUpload()
img_output = widgets.Output()
prediction_lbl = widgets.Label()
```

```python
def on_data_change(change):
    prediction_lbl = ""
    img = PILImage.create(upload_btn.data[-1])
    img_output.clear_output()
    with img_output: display(img.to_thumb(128, 128))
    prediction, prediction_idx, probabilities = learner.predict(img)
    prediction_lbl.value = f"Prediction: {prediction}, Probability: {probabilities[prediction_idx]:0.4f}"
```

```python
upload_btn.observe(on_data_change, names=["data"])
```

```python
display(VBox([widgets.Label("Select your bear!"), upload_btn, img_output, prediction_lbl]))
```
